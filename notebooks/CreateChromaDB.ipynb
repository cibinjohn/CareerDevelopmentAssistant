```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_chroma import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings
```

```python
PDF_PATH = "../data/RAG_sample_data.pdf"
CHROMA_DIR = "../data/chroma_db"
```

```python
loader = PyPDFLoader(PDF_PATH)
```

```python
documents = loader.load()
print(type(documents))
```

```python
documents[0]
```

```python
full_text = "\n".join([doc.page_content for doc in documents])
full_text[:100]
```

```python
print("Splitting by ## separator...")
raw_chunks = full_text.replace("\n"," ").split("##")
print("sample chunk : ",raw_chunks[0])
```

```python
cleaned_chunks = [chunk.strip() for chunk in raw_chunks if chunk.strip()]
cleaned_chunks
```

```python
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
# import shutil
#
# shutil.rmtree("../data/chroma_db")
# print("ChromaDB deleted.")
```

```python
vectorstore = Chroma(
        persist_directory=CHROMA_DIR,
        embedding_function=embeddings
    )
```

```python
# Add each role chunk with metadata
vectorstore.add_texts(
    texts=cleaned_chunks,
    metadatas=[
        {
            "source": PDF_PATH
        }
        for chunk in cleaned_chunks
    ]
)
```
