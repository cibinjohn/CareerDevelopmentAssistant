```python
from langchain_anthropic import ChatAnthropic
from langchain_core.messages import HumanMessage
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
ANTHROPIC_BASE_MODEL="claude-haiku-4-5-20251001"
```

```python
msg = HumanMessage(content="Hello")
msg
```

```python
# Initialize Claude
llm = ChatAnthropic(
    model=ANTHROPIC_BASE_MODEL,  # fast & cheap (good for dev)
    temperature=0.7,
    streaming=True
)
```

```python
resp = llm.stream([HumanMessage(content="how are you")])
```

```python
for chunk in resp:
    print(chunk.content)
```
