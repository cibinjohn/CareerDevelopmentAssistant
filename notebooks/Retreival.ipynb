```python
from langchain.tools import tool
from langchain_chroma import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings
```

```python
CHROMA_DIR = "../data/chroma_db"

embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
vectorstore = Chroma(
    persist_directory=CHROMA_DIR,
    embedding_function=embeddings
)
```

```python
retriever = vectorstore.as_retriever(search_kwargs={"k": 3})
```

```python
def career_rag_tool(query: str) -> str:
    """
    Search the career knowledge base for relevant information about
    IT roles such as Data Engineer, Data Analyst, Data Scientist,
    ML Engineer, Database Administrator, and Data Architect.
    Use this tool when the user asks about any of these roles,
    required skills, responsibilities, or career paths.
    """
    docs = retriever.invoke(query)
    if not docs:
        return "No relevant information found in the career knowledge base."

    results = []
    for i, doc in enumerate(docs, 1):
        # role = doc.metadata.get("role", "Unknown Role")
        results.append(f"{doc.page_content.strip()}")

    return results
```

```python
chunks=career_rag_tool("who is data scientist")
for chunk in chunks:
    print("####")
    print(chunk)
```
